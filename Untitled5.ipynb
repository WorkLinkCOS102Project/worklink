```python
import sys
import os
import tkinter as tk
from tkinter import messagebox
import ttkbootstrap as tb

# =====================================================================
# 📂 DATA MANAGER INTERFACE (Connect your data_manager.ipynb functions here)
# =====================================================================
# These dummy functions mimic database saving/loading. Replace these or 
# map them directly to the functions inside your team's data_manager!

MOCK_DATABASE = []

def data_manager_save_job(title, description, skills, experience, employer_id):
    """Saves a newly posted job opportunity to the storage system."""
    job_record = {
        "id": len(MOCK_DATABASE) + 1,
        "title": title,
        "description": description,
        "skills": skills,
        "experience": experience,
        "employer_id": employer_id
    }
    MOCK_DATABASE.append(job_record)
    return True

def data_manager_get_employer_jobs(employer_id):
    """Retrieves all jobs posted by a specific employer ID."""
    return [job for job in MOCK_DATABASE if job["employer_id"] == employer_id]

# =====================================================================
# 🖥️ WORKLINK EMPLOYER PANEL APPLICATION
# =====================================================================

class EmployerApp(tb.Window):
    def __init__(self):
        # Initializing window with a professional 'superhero' dark theme
        super().__init__(themename="superhero")
        self.title("WorkLink — Employer Control Panel")
        self.geometry("1050x650")
        
        # Simulating a logged-in employer instance for tracking data rows
        self.logged_in_employer_id = "Employer_Group6"

        self.current_screen = None
        # Start immediately on the main Employer Hub Dashboard
        self.show_screen(EmployerDashboardScreen)

    def show_screen(self, screen_class):
        """Swaps frames cleanly to simulate multi-page architecture."""
        if self.current_screen is not None:
            self.current_screen.destroy()

        self.current_screen = screen_class(self)
        self.current_screen.pack(fill="both", expand=True)


class EmployerDashboardScreen(tb.Frame):
    """The central hub for an employer to view their history and select actions."""
    def __init__(self, parent):
        super().__init__(parent)
        self.parent = parent

        # --- HEADER ---
        header_frame = tb.Frame(self, padding=20)
        header_frame.pack(fill="x")
        
        tb.Label(header_frame, text="WorkLink Employer Dashboard", font=("Segoe UI", 22, "bold"), bootstyle="primary").pack(side="left")
        tb.Label(header_frame, text=f"ID: {parent.logged_in_employer_id}", font=("Segoe UI", 10, "italic"), foreground="#9ca3af").pack(side="right", pady=10)

        # --- ACTION BAR ---
        action_bar = tb.Frame(self, padding=10)
        action_bar.pack(fill="x", padx=15)
        
        post_job_btn = tb.Button(action_bar, text="➕ Post New Job", bootstyle="success", command=lambda: parent.show_screen(PostJobFormScreen))
        post_job_btn.pack(side="left", padx=5)
        
        refresh_btn = tb.Button(action_bar, text="🔄 Refresh Listings", bootstyle="secondary-outline", command=self.load_posted_jobs)
        refresh_btn.pack(side="left", padx=5)

        # --- RECENTLY POSTED JOBS TABLE ---
        body_frame = tb.Frame(self, padding=15)
        body_frame.pack(fill="both", expand=True)
        
        tb.Label(body_frame, text="Your Currently Posted Opportunities:", font=("Segoe UI", 12, "bold")).pack(anchor="w", pady=(0, 10))

        # Table Layout UI using Treeview
        scroll_y = tb.Scrollbar(body_frame, orient="vertical")
        self.job_tree = tb.Treeview(body_frame, yscrollcommand=scroll_y.set, show="headings", bootstyle="primary")
        scroll_y.config(command=self.job_tree.yview)
        
        scroll_y.pack(side="right", fill="y")
        self.job_tree.pack(side="left", fill="both", expand=True)

        # Define table columns matching step 3 configurations
        self.job_tree["columns"] = ("ID", "Title", "Experience", "Skills Required", "Description")
        
        self.job_tree.heading("ID", text="Job ID", anchor="w")
        self.job_tree.heading("Title", text="Job Title", anchor="w")
        self.job_tree.heading("Experience", text="Experience Level", anchor="w")
        self.job_tree.heading("Skills Required", text="Required Skills", anchor="w")
        self.job_tree.heading("Description", text="Description Summary", anchor="w")

        self.job_tree.column("ID", width=70, minwidth=50, stretch=False)
        self.job_tree.column("Title", width=200, minwidth=150)
        self.job_tree.column("Experience", width=150, minwidth=120, stretch=False)
        self.job_tree.column("Skills Required", width=250, minwidth=180)
        self.job_tree.column("Description", width=300, minwidth=200)

        # Automatically pull postings from the data manager on window load
        self.load_posted_jobs()

    def load_posted_jobs(self):
        """Uses data_manager layer to read and display rows inside the view field table."""
        # Wipe old rows safely
        self.job_tree.delete(*self.job_tree.get_children())
        
        # Pull data records via data manager query
        my_jobs = data_manager_get_employer_jobs(self.parent.logged_in_employer_id)
        
        for job in my_jobs:
            self.job_tree.insert("", "end", values=(
                job["id"],
                job["title"],
                job["experience"],
                job["skills"],
                job["description"]
            ))


class PostJobFormScreen(tb.Frame):
    """The submission form screen providing input controls outlined in Step 3."""
    def __init__(self, parent):
        super().__init__(parent)
        self.parent = parent

        # --- HEADER ---
        header_frame = tb.Frame(self, padding=20)
        header_frame.pack(fill="x")
        
        back_btn = tb.Button(header_frame, text="← Back to Dashboard", bootstyle="secondary", command=lambda: parent.show_screen(EmployerDashboardScreen))
        back_btn.pack(side="left", padx=(0, 20))
        
        tb.Label(header_frame, text="Create Job Opportunity Listing", font=("Segoe UI", 18, "bold"), bootstyle="success").pack(side="left")

        # --- SCROLLABLE FORM PANEL CONTAINER ---
        form_container = tb.Frame(self, padding=30)
        form_container.pack(fill="both", expand=True)

        # 1. Job Title Field Entry
        tb.Label(form_container, text="Job Title *", font=("Segoe UI", 11, "bold")).pack(anchor="w", pady=(0, 5))
        self.title_entry = tb.Entry(form_container, width=60)
        self.title_entry.pack(anchor="w", pady=(0, 20))
        self.title_entry.placeholder_text = "e.g., Junior Python Developer, Assistant Project Manager"

        # 2. Experience Level Dropdown (Incorporates Optional Entry Level Focus Points)
        tb.Label(form_container, text="Experience Requirements Level *", font=("Segoe UI", 11, "bold")).pack(anchor="w", pady=(0, 5))
        self.experience_var = tk.StringVar()
        self.experience_dropdown = tb.Combobox(
            form_container, 
            textvariable=self.experience_var, 
            values=["No Experience Required (Entry-Level)", "Intermediate (1-3 Years)", "Senior (5+ Years)"],
            state="readonly",
            width=58
        )
        self.experience_dropdown.pack(anchor="w", pady=(0, 20))
        self.experience_dropdown.current(0) # Defaults to Entry Level to optimize unemployment solution metrics

        # 3. Skills Required Entry
        tb.Label(form_container, text="Required Skills Base (Separated by commas) *", font=("Segoe UI", 11, "bold")).pack(anchor="w", pady=(0, 5))
        self.skills_entry = tb.Entry(form_container, width=60)
        self.skills_entry.pack(anchor="w", pady=(0, 20))

        # 4. Job Description Text Box Panel
        tb.Label(form_container, text="Detailed Job Description *", font=("Segoe UI", 11, "bold")).pack(anchor="w", pady=(0, 5))
        self.description_text = tb.Text(form_container, height=6, width=70, font=("Segoe UI", 10))
        self.description_text.pack(anchor="w", pady=(0, 25))

        # --- SUBMIT FRAME SYSTEM BUTTON ---
        submit_btn = tb.Button(
            form_container, 
            text="🚀 Save and Publish Posting", 
            bootstyle="success", 
            padding=12,
            command=self.process_job_submission
        )
        submit_btn.pack(anchor="w")

    def process_job_submission(self):
        """Validates all input entries and commits transaction cleanly via the Data Manager."""
        title = self.title_entry.get().strip()
        experience = self.experience_var.get()
        skills = self.skills_entry.get().strip()
        description = self.description_text.get("1.0", tk.END).strip()
        employer = self.parent.logged_in_employer_id

        # Structural UI form validation validations
        if not title or not skills or not description:
            messagebox.showwarning("Validation Error", "Please complete all mandatory form fields before saving.")
            return

        # Fire operation up into your data_manager layer 
        success = data_manager_save_job(title, description, skills, experience, employer)

        if success:
            messagebox.showinfo("Job Saved", f"Success! '{title}' has been written to the data management system and is now live for Job Seekers.")
            # Route employer right back onto their overview grid layout board
            self.parent.show_screen(EmployerDashboardScreen)
        else:
            messagebox.showerror("System Failure", "The file manager database was unable to save this job data entry tracking record row.")


if __name__ == "__main__":
    app = EmployerApp()
    app.mainloop()
```
