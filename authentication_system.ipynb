```python
import tkinter as tk
from tkinter import messagebox
from ipynb.fs.full.data_manager import initialise_file, save_user, validate_login

initialise_file()

root = tk.Tk()
root.title("WorkLink - Welcome")
root.geometry("500x600")
root.config(bg="#f0f0f0")

def clear_screen():
    for widget in root.winfo_children():
        widget.destroy()

def role_selection_screen():
    clear_screen()
    root.geometry("500x500")

    title = tk.Label(root, text="Welcome to WorkLink", font=("Arial", 24, "bold"), bg="#f0f0f0", fg="black")
    title.pack(pady=40)
    
    subtitle = tk.Label(root, text="Please select your role to continue:", font=("Arial", 12), bg="#f0f0f0", fg="gray")
    subtitle.pack(pady=10)

    employer_btn = tk.Button(
        root, text="I am an Employer", font=("Arial", 12, "bold"), 
        bg="blue", fg="white", width=25, height=2,
        command=lambda: login_screen("Employer")
    )
    employer_btn.pack(pady=15)

    seeker_btn = tk.Button(
        root, text="I am a Job Seeker", font=("Arial", 12, "bold"), 
        bg="green", fg="white", width=25, height=2,
        command=lambda: login_screen("Job Seeker")
    )
    seeker_btn.pack(pady=15)

    exit_btn = tk.Button(root, text="Exit", font=("Arial", 10), bg="gray", fg="white", width=10, command=root.destroy)
    exit_btn.pack(pady=30)

def login_screen(role):
    clear_screen()
    root.geometry("500x550")

    title = tk.Label(root, text=f"{role} Login", font=("Arial", 20, "bold"), bg="#f0f0f0", fg="black")
    title.pack(pady=30)

    tk.Label(root, text="Email Address", font=("Arial", 11), bg="#f0f0f0").pack(anchor="w", padx=100)
    email_entry = tk.Entry(root, width=35, font=("Arial", 11), relief="solid", bd=1)
    email_entry.pack(pady=5)

    tk.Label(root, text="Password", font=("Arial", 11), bg="#f0f0f0").pack(anchor="w", padx=100)
    password_entry = tk.Entry(root, width=35, font=("Arial", 11), show="*", relief="solid", bd=1)
    password_entry.pack(pady=5)

    def handle_login():
        email = email_entry.get().strip()
        password = password_entry.get().strip()

        if not email or not password:
            messagebox.showerror("Error", "Please fill in all fields.")
            return

        user_profile = validate_login(email, password)

        if user_profile:
            if user_profile["role"] != role:
                messagebox.showerror("Role Mismatch", f"This account is registered as a {user_profile['role']}, not a {role}.")
                return
            
            messagebox.showinfo("Success", f"Welcome back, {user_profile['name']}!")
            
            if role == "Employer":
                launch_employer_system(user_profile["id"], user_profile["name"])
            else:
                launch_job_seeker_system(user_profile["id"], user_profile["name"])
        else:
            messagebox.showerror("Error", "Invalid email or password.")

    submit_btn = tk.Button(root, text="Login", font=("Arial", 12, "bold"), bg="blue" if role == "Employer" else "green", fg="white", width=20, command=handle_login)
    submit_btn.pack(pady=20)

    register_lnk = tk.Button(root, text="Don't have an account? Register here", font=("Arial", 10, "underline"), bg="#f0f0f0", fg="black", bd=0, activebackground="#f0f0f0", command=lambda: register_screen(role))
    register_lnk.pack(pady=10)

    back_btn = tk.Button(root, text="Back to Roles", font=("Arial", 10), bg="gray", fg="white", width=15, command=role_selection_screen)
    back_btn.pack(pady=15)

def register_screen(role):
    clear_screen()
    root.geometry("500x600")

    title = tk.Label(root, text=f"Register as {role}", font=("Arial", 20, "bold"), bg="#f0f0f0", fg="black")
    title.pack(pady=25)

    tk.Label(root, text="Full Name / Company Name", font=("Arial", 11), bg="#f0f0f0").pack(anchor="w", padx=100)
    name_entry = tk.Entry(root, width=35, font=("Arial", 11), relief="solid", bd=1)
    name_entry.pack(pady=5)

    tk.Label(root, text="Email Address", font=("Arial", 11), bg="#f0f0f0").pack(anchor="w", padx=100)
    email_entry = tk.Entry(root, width=35, font=("Arial", 11), relief="solid", bd=1)
    email_entry.pack(pady=5)

    tk.Label(root, text="Password", font=("Arial", 11), bg="#f0f0f0").pack(anchor="w", padx=100)
    password_entry = tk.Entry(root, width=35, font=("Arial", 11), show="*", relief="solid", bd=1)
    password_entry.pack(pady=5)

    skills_label_text = "Company Details / Industry" if role == "Employer" else "Your Skills (e.g. Python, Typing)"
    tk.Label(root, text=skills_label_text, font=("Arial", 11), bg="#f0f0f0").pack(anchor="w", padx=100)
    skills_entry = tk.Entry(root, width=35, font=("Arial", 11), relief="solid", bd=1)
    skills_entry.pack(pady=5)

    def handle_register():
        name = name_entry.get().strip()
        email = email_entry.get().strip()
        password = password_entry.get().strip()
        skills = skills_entry.get().strip()

        if not name or not email or not password or not skills:
            messagebox.showerror("Error", "All fields are required.")
            return

        success = save_user(name, email, password, role, skills)

        if success:
            messagebox.showinfo("Success", "Account created successfully! Please log in.")
            login_screen(role)
        else:
            messagebox.showerror("Error", "An account with this email already exists.")

    submit_btn = tk.Button(root, text="Create Account", font=("Arial", 12, "bold"), bg="blue" if role == "Employer" else "green", fg="white", width=20, command=handle_register)
    submit_btn.pack(pady=20)

    login_lnk = tk.Button(root, text="Already have an account? Login here", font=("Arial", 10, "underline"), bg="#f0f0f0", fg="black", bd=0, activebackground="#f0f0f0", command=lambda: login_screen(role))
    login_lnk.pack(pady=10)

def launch_employer_system(emp_id, emp_name):
    clear_screen()

    lbl = tk.Label(root, text=f"Handoff to Employer App\nID: {emp_id}\nUser: {emp_name}", font=("Arial", 16), bg="#f0f0f0")
    lbl.pack(pady=100)

def launch_job_seeker_system(user_id, user_name):
    clear_screen()
    lbl = tk.Label(root, text=f"Handoff to Job Seeker App\nID: {user_id}\nUser: {user_name}", font=("Arial", 16), bg="#f0f0f0")
    lbl.pack(pady=100)

role_selection_screen()
root.mainloop()
```
